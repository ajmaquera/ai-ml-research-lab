```python
import numpy as np

X = np.array(([0,0],
              [0,1],
              [1,0],
              [1,1]), dtype=float)
y = np.array(([0],
              [1],
              [1],
              [0]), dtype=float)
```

```python
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.optimizers import SGD   # SGD: Stochastic Gradient Descent

model = keras.Sequential([
    layers.Dense(4, activation='tanh', input_shape=(2,)),
    layers.Dense(1, activation='tanh')
])

sgd = SGD(learning_rate=0.1)
model.compile(optimizer=sgd,
              loss='mean_squared_error')

model.fit(X, y, epochs=500, batch_size=1)
model.evaluate(X, y)
```

```python
y_hat = model.predict(X).round()
y_hat
```

```python
x_0 = np.linspace(0,1,100)
x_1 = np.linspace(0,1,100)
```

```python
points = []
for _x0 in x_0:
  for _x1 in x_1:
    points.append([_x0, _x1])

points = np.array(points)
X_0, X_1 = np.meshgrid(x_0, x_1)
```

```python
_y_hat = model.predict(points)
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(6,6))
ax.contourf(X_0, X_1, _y_hat.reshape(100,100), cmap='plasma')
plt.show()
```
