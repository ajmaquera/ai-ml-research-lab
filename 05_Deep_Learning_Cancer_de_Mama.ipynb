```python
import pandas as pd

# Data de la universidad de Wisconsin
df = pd.read_csv('/content/drive/MyDrive/Machine Learning/Data/data_cancer.csv', index_col=0)
df.head()
```

```python
# Eliminamos la ultima columna
df = df.drop("Unnamed: 32",axis=1)

# Mezclamos todos los datos a fin de garantizar un buen funcionamiento del modelo
df = df.sample(frac=1)

y = df.diagnosis
X = df.drop("diagnosis",axis=1)
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
X = scaler.fit_transform(X)
```

```python
y = y.replace(["M","B"],[1,0])
```

```python
X,y
```

```python
split = round(X.shape[0]*0.7)
X_train, X_test = X[:split], X[split:]
y_train, y_test = y[:split], y[split:]
```

```python
print("Tamaños de X_train", X_train.shape)
print("Tamaños de X_test", X_test.shape)
print("Tamaños de Y_train", y_train.shape)
print("Tamaños de Y_test", y_test.shape)
```

```python
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.optimizers import SGD

model = keras.Sequential([
    layers.Dense(20, activation='relu',input_shape=(30,)),  # Es 30 porque tenemos 30 variables predictoras
    layers.Dense(20, activation='relu'),
    layers.Dense(20, activation='relu'),
    layers.Dense(1, activation='sigmoid')   # porque la salida es binaria y es recomendable 'sigmoid'
])

sgd = SGD(learning_rate = 0.1)
model.compile(optimizer=sgd,
              loss='mean_squared_error')

model.fit(X_train, y_train, epochs=500, batch_size=10)
```

```python
y_hat_train = model.predict(X_train).round()

y_hat_train[:15]
```

```python
y_hat_test = model.predict(X_test).round()
y_hat_test[:15]
```

```python
from sklearn.metrics import confusion_matrix
confusion_matrix(y_test, y_hat_test)
```
